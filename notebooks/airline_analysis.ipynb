```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("airline_passenger_data.csv")

print(df.head())
print(df.info())
```

```python
# check missing values
print(df.isnull().sum())

# convert date
df["Departure Date"] = pd.to_datetime(df["Departure Date"])

# remove duplicates
df = df.drop_duplicates()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df["Age"], bins=20)
plt.title("Passenger Age Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Flight Status", data=df)
plt.title("Flight Status Overview")
plt.show()
```

```python
top_airports = df["Airport Name"].value_counts().head(10)

top_airports.plot(kind="bar")
plt.title("Top Airports by Passenger Count")
plt.show()
```

```python
top_nations = df["Nationality"].value_counts().head(10)

sns.barplot(x=top_nations.values, y=top_nations.index)
plt.title("Top Passenger Nationalities")
plt.show()
```
